### data injection

```python
from langchain_core.documents import Document

doc_example = Document(page_content="Hello world", metadata={"source": "test.txt"})
print(doc_example)
```

```python
from langchain_community.document_loaders import TextLoader

text_loader = TextLoader("../data/txt/animal-fact.txt", encoding="utf-8")
text_document = text_loader.load()
text_document
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader

dir_pdf_loader = DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf",
    loader_cls=PyPDFLoader,
)
pdf_document = dir_pdf_loader.load()
pdf_document
```
